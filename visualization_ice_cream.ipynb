```python
import pandas as pd

df_per = pd.read_csv("perek_morozhenoe.csv")
df_vv = pd.read_csv("vkusvill_morozhenoe.csv")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

top25 = df_per.sort_values("Score", ascending=False).head(25).reset_index(drop=True)


palette = sns.color_palette("husl", n_colors=len(top25))


fig, ax = plt.subplots(figsize=(11, 9))

for i, row in top25.iterrows():
    ax.scatter(
        row["Относительная популярность"],
        row["Ценовая премия"],
        s=200,
        color=palette[i],
        edgecolor="white",
        linewidth=1.5,
        label=f"{i+1}. {row['Название'][:50]} (score {row['Score']:.2f})",
        zorder=3,
    )

    ax.annotate(
        str(i + 1),
        (row["Относительная популярность"], row["Ценовая премия"]),
        xytext=(7, 7),
        textcoords="offset points",
        fontsize=10,
        fontweight="bold",
        color="#333333",
        zorder=4,
    )

ax.axhline(1, color="black", linestyle="--", alpha=0.4, linewidth=1)
ax.axvline(1, color="black", linestyle="--", alpha=0.4, linewidth=1)

ax.set_xlabel("Относительная популярность (>1 = популярнее медианы ниши)", fontsize=11)
ax.set_ylabel("Премия по цене (>1 = дороже медианы ниши)", fontsize=11)
ax.set_title("Топ-25 кандидатов", fontsize=14, fontweight="bold", pad=15)

ax.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.10),
    ncol=2,
    fontsize=9,
    frameon=False,
)

sns.despine()
plt.tight_layout()
plt.show()
```

```python
sorbet_per = df_per[
    df_per["Название"].str.contains("сорбет|Сорбет", case=False, na=False)
].reset_index(drop=True)
sorbet_vv = df_vv[
    df_vv["Название"].str.contains("сорбет|Сорбет", case=False, na=False)
].reset_index(drop=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Сорбеты Перекрёстка

sns.set_style("whitegrid")

top_sp = sorbet_per.sort_values("Score", ascending=False).reset_index(drop=True)

palette = sns.color_palette("husl", n_colors=len(top_sp))

fig, ax = plt.subplots(figsize=(11, 9))

for i, row in top_sp.iterrows():
    is_star = "малина-личи" in row["Название"].lower()

    ax.scatter(
        row["Относительная популярность"],
        row["Ценовая премия"],
        s=350 if is_star else 200,
        color="#E91E8C" if is_star else palette[i],
        edgecolor="#880E4F" if is_star else "white",
        linewidth=2.5 if is_star else 1.5,
        label=f"{'★ ' if is_star else ''}{i+1}. {row['Название'][:50]} (score {row['Score']:.2f})",
        zorder=5 if is_star else 3,
    )

    ax.annotate(
        "★" if is_star else str(i + 1),
        (row["Относительная популярность"], row["Ценовая премия"]),
        xytext=(7, 7),
        textcoords="offset points",
        fontsize=11 if is_star else 10,
        fontweight="bold",
        color="#E91E8C" if is_star else "#333333",
        zorder=6,
    )

ax.axhline(1, color="black", linestyle="--", alpha=0.4, linewidth=1)
ax.axvline(1, color="black", linestyle="--", alpha=0.4, linewidth=1)

ax.set_xlabel("Относительная популярность (>1 = популярнее медианы ниши)", fontsize=11)
ax.set_ylabel("Премия по цене (>1 = дороже медианы ниши)", fontsize=11)
ax.set_title("Сорбеты — Перекрёсток", fontsize=14, fontweight="bold", pad=15)

ax.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.10),
    ncol=2,
    fontsize=9,
    frameon=False,
)

sns.despine()
plt.tight_layout()
plt.show()

# Сорбеты ВкусВилла
sns.set_style("whitegrid")

top_sv = sorbet_vv.sort_values("score", ascending=False).reset_index(drop=True)

palette = sns.color_palette("husl", n_colors=len(top_sv))

fig, ax = plt.subplots(figsize=(11, 9))

for i, row in top_sv.iterrows():
    is_star = row["Название"].strip() == 'Сорбет "Малина-Личи"'

    ax.scatter(
        row["Относит. популярность"],
        row["Премия по цене"],
        s=350 if is_star else 200,
        color="#E91E8C" if is_star else palette[i],
        edgecolor="#880E4F" if is_star else "white",
        linewidth=2.5 if is_star else 1.5,
        label=f"{'★ ' if is_star else ''}{i+1}. {row['Название'][:50]} (score {row['score']:.2f})",
        zorder=5 if is_star else 3,
    )

    ax.annotate(
        "★" if is_star else str(i + 1),
        (row["Относит. популярность"], row["Премия по цене"]),
        xytext=(7, 7),
        textcoords="offset points",
        fontsize=11 if is_star else 10,
        fontweight="bold",
        color="#E91E8C" if is_star else "#333333",
        zorder=6,
    )

ax.axhline(1, color="black", linestyle="--", alpha=0.4, linewidth=1)
ax.axvline(1, color="black", linestyle="--", alpha=0.4, linewidth=1)

ax.set_xlabel("Относительная популярность (>1 = популярнее медианы ниши)", fontsize=11)
ax.set_ylabel("Премия по цене (>1 = дороже медианы ниши)", fontsize=11)
ax.set_title("Сорбеты — ВкусВилл", fontsize=14, fontweight="bold", pad=15)

ax.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.10),
    ncol=2,
    fontsize=9,
    frameon=False,
)

sns.despine()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# Цена Сорбет "Малина-Личи" из ВкусВилла
ml_price_vv = sorbet_vv.loc[
    sorbet_vv["Название"].str.strip() == 'Сорбет "Малина-Личи"', "Цена за 100 г, ₽"
].iloc[0]

# Цена Малина-Личи в Перекрёстке (300г, Зелёная Линия)
ml_price_per = sorbet_per.loc[
    sorbet_per["Название"].str.contains("Малина-Личи Зелёная Линия, 300г", na=False),
    "Цена за 100 г, ₽",
].iloc[0]


def hist_price(df, title, color, ml_price):
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.hist(
        df["Цена за 100 г, ₽"].dropna(),
        bins=20,
        color=color,
        edgecolor="white",
        linewidth=1.2,
    )

    median = df["Цена за 100 г, ₽"].median()
    ymax = ax.get_ylim()[1]

    ax.axvline(median, color="black", linestyle="--", alpha=0.6, linewidth=1.2)
    ax.text(
        median,
        ymax * 0.95,
        f" медиана = {median:.0f} ₽",
        fontsize=9,
        color="#333",
        va="top",
    )

    # Малина-Личи
    ax.axvline(ml_price, color="#E91E8C", linestyle="-", linewidth=2.2, alpha=0.9)
    ax.text(
        ml_price,
        ymax * 0.80,
        f" ★ Малина-Личи = {ml_price:.0f} ₽",
        fontsize=10,
        color="#E91E8C",
        va="top",
        fontweight="bold",
    )

    ax.set_xlabel("Цена за 100 г, ₽", fontsize=11)
    ax.set_ylabel("Кол-во позиций", fontsize=11)
    ax.set_title(title, fontsize=13, fontweight="bold", pad=12)
    sns.despine()
    plt.tight_layout()
    plt.show()


# Всё мороженое — Перекрёсток (цена Малина-Личи Зелёной Линии 300г)
hist_price(
    df_per, "Распределение цены — всё мороженое, Перекрёсток", "#5C6BC0", ml_price_per
)

# Всё мороженое — ВкусВилл
hist_price(
    df_vv, "Распределение цены — всё мороженое, ВкусВилл", "#26A69A", ml_price_vv
)
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns

sns.set_style("whitegrid")

STAR = "#E91E8C"
STAR_EDGE = "#880E4F"
PER_COLOR = "#5C6BC0"
VV_COLOR = "#26A69A"


def is_star(name):
    return "малина-личи" in str(name).lower()


fig, ax = plt.subplots(figsize=(11, 7))

for df_s, label, color, marker in [
    (sorbet_per, "Перекрёсток", PER_COLOR, "o"),
    (sorbet_vv, "ВкусВилл", VV_COLOR, "s"),
]:
    for _, row in df_s.iterrows():
        star = is_star(row["Название"])
        ax.scatter(
            row["Кол-во отзывов"],
            row["Рейтинг"],
            s=400 if star else 130,
            color=STAR if star else color,
            edgecolor=STAR_EDGE if star else "white",
            linewidth=2.5 if star else 1,
            marker="*" if star else marker,
            zorder=6 if star else 3,
            alpha=1.0 if star else 0.75,
        )
        if star:
            ax.annotate(
                f"★ Малина-Личи ({label})",
                (row["Кол-во отзывов"], row["Рейтинг"]),
                xytext=(12, 8),
                textcoords="offset points",
                fontsize=10,
                color=STAR,
                fontweight="bold",
            )

handles = [
    mpatches.Patch(color=PER_COLOR, label="Перекрёсток"),
    mpatches.Patch(color=VV_COLOR, label="ВкусВилл"),
    mpatches.Patch(color=STAR, label="★ Малина-Личи"),
]
ax.legend(handles=handles, fontsize=10, frameon=True, loc="lower right")
ax.set_xlabel("Кол-во отзывов", fontsize=11)
ax.set_ylabel("Рейтинг", fontsize=11)
ax.set_title(
    "Сорбеты: отзывы × рейтинг — оба магазина", fontsize=13, fontweight="bold", pad=12
)
sns.despine()
plt.tight_layout()
plt.show()
```
